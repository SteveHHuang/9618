```python
#insertion sort
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key

arr = [5, 2, 4, 6, 1, 3]
insertion_sort(arr)
print(arr)
```

```python
import random

# extension_selectionsort
def selectionsort(arr):
    for i in range(len(arr)):
        lowest = i 
        for j in range(i,len(arr)):
            if arr[j] < arr[lowest]:
                lowest = j
        arr[i],arr[lowest] = arr[lowest], arr[i]
            
            
arr = [random.randint(1,100) for k in range (10)]

print(arr)

selectionsort(arr)
print(arr)
```
