```python
#insertion sort
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key

arr = [5, 2, 4, 6, 1, 3]
insertion_sort(arr)
print(arr)
```

```python
# bubble sort
def bubble_sort(arr):
    j = len(arr)        
    while True:
        swap = False
        for k in range (j-1):
            if arr[k+1] < arr[k]:
                temp = arr[k+1]
                arr[k+1] = arr[k]
                arr[k] = temp
                swap = True
        j -= 1
        if swap == False:
            break 

bubble_sort(arr)        
print(arr)
```
