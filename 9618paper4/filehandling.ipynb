```python
# 20250610
#file handling
#read, write, append

#read txt file
set7 = "dataset7.txt"
set6 = "dataset6.txt"
f = open(set6)
#content0 = f.read() # 一次性读取文件的所有内容
content1 = f.readlines(23) 
# 读取文件的n个字节 并把每一行的内容以数组的形式返回 用\n来分隔每一行的内容 没有参数的时候默认读取所有的内容

content2 = f.readline(6) # 读取文件一行内容中前n个字符 每遇到一次空格后自动换行输出
content3 = f.readline(6)
content0 = f.read() 

print(content1)
print(content2)
print(content3)
print(content0)
f.close() #关闭文件 一定要写！！！不然会扣分
# 每次操作之后的指针不会回到第一行 会停留到上一次操作之后的位置
```
