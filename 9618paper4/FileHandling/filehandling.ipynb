```python
arr = []
with open("dataset8.txt","r") as g: # read only; 'w': write file
    while True:
        temp = g.readline(3)
        if len(temp) != 0:
            arr.append(int(temp))
        else:
            break
g.close()
print(arr)
```

```python
# read data, using bubble sort, timing code
import time,random
numbers=[]
# read data from txt file
f=open("dataset6.txt")
for line in f:
    numbers.append((int(line.strip()))) 

# strip() 会删除字符串两侧的空白字符（包括空格、换行符 \n、制表符 \t 等）
# spilt方法默认隔开每个字符 split() 的默认行为是按空白字符分割字符串​（任意长度的空格/换行等）。
# 如果指定分隔符（如 split(' ')），会按指定字符分割。

# start High-precision timing, bubble sort
start = time.perf_counter()  
for i in range(len(numbers)-1):
    for j in range(len(numbers)-i-1):
        if numbers[j]>numbers[j+1]:
            temp=numbers[j]
            numbers[j]=numbers[j+1]
            numbers[j+1]=temp
# end timing 
end = time.perf_counter()
print(f"time consumption: {end - start:.6f} second")# Format with 6 decimal places
print(numbers)
```
