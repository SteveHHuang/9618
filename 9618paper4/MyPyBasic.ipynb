```python
# input
# input a number, sum all non negative numbers, negative to exit
num = int(input("please input a number, negative to exit")) # input data is regarded as a type of string in python 
sum = 0
while num >= 0:
    sum += num
    num = int(input("please input a number, negative to exit"))
print("sum is", sum)
```
