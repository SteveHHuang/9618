```python
# read data, using bubble sort, timing code
import time,random
numbers=[]
# read data from txt file
f=open("dataset3.txt")
for line in f:
    numbers.append((int)(line.strip()))
# start High-precision timing, bubble sort
start = time.perf_counter()  
for i in range(len(numbers)-1):
    for j in range(len(numbers)-i-1):
        if numbers[j]<numbers[j+1]:
            temp=numbers[j]
            numbers[j]=numbers[j+1]
            numbers[j+1]=temp
# end timing
end = time.perf_counter()
print(f"time consumption: {end - start:.6f} second")  # Format with 6 decimal places
```

```python
# read data, using insertion sort,timing code
import time,random
# read data in numbers from txt file
numbers=[]
f=open("dataset3.txt")
for line in f:
    numbers.append((int)(line.strip()))

# start High-precision timing, insertion sort
start = time.perf_counter()  
for i in range(1,len(numbers)):
    j=i-1
    temp=numbers[i]
    while j>=0 and temp>numbers[j]:
        numbers[j+1]=numbers[j]
        j-=1
    numbers[j+1]=temp
end = time.perf_counter()
print(f"time consumption: {end - start:.6f} second")  # Format with 6 decimal places
```

```python
#efficient bubble sort
import time,random
numbers=[]
f=open("dataset2.txt")
for line in f:
    numbers.append((int)(line.strip()))
swap=True
start = time.perf_counter()
while swap==True:
    swap=False
    index=len(numbers)-1
    for i in range(index):
        if numbers[i]>numbers[i+1]:
            swap=True
            temp=numbers[i]
            numbers[i]=numbers[i+1]
            numbers[i+1]=temp
    index-=1
end = time.perf_counter()
print(f"time consumption: {end - start:.6f} second")  # Format with 6 decimal places
```
