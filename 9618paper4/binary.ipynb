```python
import random 

arr = [random.randint(1,1000) for i in range(10)]
```

```python
# bubble sort
for i in range(10):
    for j in range(len(arr)-1):
        if arr[j] < arr[j+1]:
            temp = arr[j+1]
            arr[j+1] = arr[j]
            arr[j] = temp


print(arr)
```

```python
# insertion sort

for i in range(len(arr)):
    temp = arr[i]
    j = i-1
    while temp > arr[j] and j>=0:
        arr[j+1] = arr[j]
        j -= 1    
    arr[j+1] = temp
            
print(arr)    
```

```python
arr.sort(reverse=False)
print(arr)
```

```python
def binarySearch(num):
    global arr
    
    lowerbound = 0
    upperbound = len(arr)-1
    
    while lowerbound <= upperbound:
        mid = (upperbound+lowerbound) // 2
        if arr[mid] == num:
            return True
        elif num < arr[mid]:
            upperbound = mid-1
        elif num > arr[mid]:
            lowerbound = mid+1
        
    return False

print(binarySearch(416))

```
