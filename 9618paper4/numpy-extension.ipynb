```python
import numpy as np
nparr1 = np.array([1, 2, 3, 4], dtype='float64')

print(nparr1)
print(nparr1.shape)

nparr2 = np.array([range(i, i + 3) for i in [2, 4, 6]]) 

print(nparr2)
print(nparr2.shape)

arr_3d = np.array([[[k for k in range(j, j + 4)] for j in range(i, i + 3)] for i in [0, 2]])
print(arr_3d)
print(arr_3d.shape)  # 输出 (2, 3, 4)
```
