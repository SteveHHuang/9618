```python
class Node:
    def __init__(self,data):
        self.LeftNode = None
        self.Data = data
        self.RightNode = None
```

```python
node1 = Node(12)

node2 = Node(200)

node1.RightNode = node2

node3 = Node(10)

node1.LeftNode = node3

root = node1
```

```python
print(f"{node1.Data}({node1.LeftNode})({node1.RightNode})")
```

```python
def PrintBinTree_1(root): #中序遍历二叉树
    if root != None:
        PrintBinTree_1(root.LeftNode)
        print(root.Data)
        PrintBinTree_1(root.RightNode)
        
```

```python
def searchTree(data,ptr): 

    if ptr == None:
        return -1
    if data == ptr.Data:
        return ptr
    elif ptr.Data<data:
        return searchTree(data, ptr.RightNode)
    else:
        return searchTree(data, ptr.LeftNode)
```

```python
searchTree(10,root)
```

```python
searchTree(201,root)
```

```python
def addNode(data): #iterative adding nodes
    global root
    #比当前node小或者等于node的数放左边，否则放右边。
    if root == None:
        root = Node(data)
        return "Successfully added"
    ptr = root
    
    while True:

        if data > ptr.Data:
            if ptr.RightNode == None: 
                ptr.RightNode = Node(data)
                return "Successfully added"
            ptr = ptr.RightNode
        else: 
            if ptr.LeftNode == None: 
                ptr.LeftNode = Node(data)
                return "Successfully added"
            ptr = ptr.LeftNode
        
 
addNode(200)
addNode(239)
```

```python
def recursive_addNode(data,ptr):
    #比当前node小或者等于node的数放左边，否则放右边。
    if ptr == None:
        
        return Node(data)
    
    if ptr.Data<data:
        ptr.RightNode = recursive_addNode(data, ptr.RightNode)
    else:
        ptr.LeftNode = recursive_addNode(data, ptr.LeftNode)
    
    return ptr
```
