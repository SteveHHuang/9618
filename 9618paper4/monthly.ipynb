```python
class Candy:
    def __init__(self,flavour):
        self.__Flavour = flavour
    
    def GetFlavour(self):
        return self.__Flavour
    
```

```python
class BoxOfCandy:
    def __init__(self,row,column):
        self.CandyBox = [(None for i in range(row)) for j in range(column)]
        
    
```

```python
Box1 = BoxOfCandy(4,4)
```

```python
def moveCandyToFirstRow(col):
    global Box1
    if Box1.CandyBox[0][col] is not None:
            return True 
    for i in range(1,len(Box1.CandyBox)):
        if Box1.CandyBox[i][col] is not None:
            temp = Box1.CandyBox[i][col]
            Box1.CandyBox[0][col] = temp
            Box1.CandyBox[i][col] = None
            return True

    return False
```

```python
def removeNextByFlavor(Name):
    global Box1
    for i in range(len(Box1.CandyBox)-1, -1, -1):
        for j in range(len(Box1.CandyBox[i])):
            if Box1.CandyBox[i][j].GetFlavour() == Name:
                temp = Box1.CandyBox[i][j]
                Box1.CandyBox[i][j] = None
                return temp
            
            
    return None
            
```
