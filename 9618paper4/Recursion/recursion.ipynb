```python
#20250905
# What is recursion
# i.Function call itself
# ii.Has a base case
# iii.General case works towards the case
def fibnaccisummation(n):
    if n ==1: return n
    return n*fibnaccisummation(n-1)

print(fibnaccisummation(4))
```

```python
def pow2(base, p):
    r = 1
    for i in range(1, p+1):
        r = r*base
    return r

def pow2_recursion(b,p):
    if p <= 0: return 1
    else: return b*pow2_recursion(b,p-1)
        
```

```python
pow2_recursion(2,2)
```

```python
def fibnacci(n):
    if n <=1: return n
    return fibnacci(n-1)+fibnacci(n-2)

print(fibnacci(4))
```

```python
def iterative_factorial(n):
    result =1
    for i in range(n,0):
        result *= n
    return result

def recursive_factorial(k):
    if k <= 1: return 1
    else: return k*recursive_factorial(k-1)
    
print(recursive_factorial(10))
```

```python
# Recursive binary search

def recursiveBinarysearch(arr, target, lower, upper):
    if upper < lower:
        return -1
    
    mid = (upper+lower) //2
    
    if arr[mid] == target: return mid
    elif arr[mid] < target: return recursiveBinarysearch(arr, target, mid+1, upper)
    else: return recursiveBinarysearch(arr, target, lower, mid-1)
    
    
    
arr = [i for i in range(10)]

print(recursiveBinarysearch(arr, 6, 0, len(arr)-1))
```

```python
#Time elapse of recursive fibnacci
import time

# start High-precision timing, bubble sort
start = time.perf_counter()  
def fibnacci(n):
    if n <=1: return n
    return fibnacci(n-1)+fibnacci(n-2)

print(fibnacci(4))
end = time.perf_counter()

print(f"time consumption: {end - start:.6f} second")# Format with 6 decimal places
```
