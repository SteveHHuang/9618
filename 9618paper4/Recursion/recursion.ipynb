```python
#20250905
# What is recursion
# i.Function call itself
# ii.Has a base case
# iii.General case works towards the case
def fibnaccisummation(n):
    if n ==1: return n
    return n*fibnaccisummation(n-1)

print(fibnaccisummation(4))
```

```python
def pow2(base, p):
    r = 1
    for i in range(1, p+1):
        r = r*base
    return r

def pow2_recursion(b,p):
    if p <= 0: return 1
    else: return b*pow2_recursion(b,p-1)
        
```

```python
pow2_recursion(2,2)
```

```python
def fibnacci(n):
    if n <=1: return n
    return fibnacci(n-1)+fibnacci(n-2)

print(fibnacci(40))
```

```python
def factorial(k):
    if k <= 1: return 1
    else: return k*factorial(k-1)
    
print(factorial(10))
```
