```python
class Peson:
  def __init__(self, name, age): # 成员函数中参数最前面必须加self
    self.name = name
    self.age = age
      

p1 = Peson("John", 36)

class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age
    
  def output(): # 通过类(静态方法)调用的时候并且没有成员变量的参与才不用加self
      print("1")
    
  def __str__(self):
    return f"{self.name}({self.age})" # F string

p2 = Person("John", 36)

print(p1) 
print(p2)
Person.output()
```
