```python
#9608 w18 41 Q5
class Cards:
    def __init__(self,num,shape):
        if (num >=0 or num <= 9) and (shape == "square" or shape == "triangle" or shape == "circle"):   
            self.__Number = int(num) # self.__Number is a integer
            self.__Shape = shape # self.__Shape is a string
        else:
            print("Invalid data")
        
        
    
    def getNumber(self):
        return self.__Number
    
    def getShape(self):
        return self.__Shape
    
```

```python
#9608 w18 41 Q5
OneS = Cards(1,"square")
```

```python
#9608 w18 41 Q5
def Compare(Card1, Card2):
    Number1 = Card1.getNumber()
    Shape1 = Card1.getShape()
    Number2 = Card2.getNumber()
    Shape2 = Card2.getShape()
    if Number1 == Number2 and Shape1 == Shape2:
        print("SNAP")
        return -1
    else:
        return max(Number1,Number2)
```
