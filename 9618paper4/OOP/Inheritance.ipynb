```python
# inheritance of object 20250922

class Person: #父类，大家都能用
    def __init__(self, Name, age):
        self.__Name = Name
        self.__Age = age
    def GetName(self):
        return self.__Name
    def SetName(self,name):
        self.__Name = name
    def GetAge(self):
        return self.__Age
    def SetName(self,age):
        self.__Age = age
    def move(self):
        print("Person Move")
    def say(self):
        s = input()
        print(f"Person said: {s}")
    
```

```python
class Person1:
    def __init__(self, Name, age):
        self.Name = Name
        self.__Age = age
    def GetName(self):
        return self.__Name
    def SetName(self,name):
        self.__Name  = name
    def GetAge(self):
        return self.__Age
    def SetName(self,age):
        self.__Age  = age
    def move(self):
        print("Person Move")
    def say(s):
        s = input()
        print(f"Person said: {s}")
    
    @classmethod #indep of instaces
    def countClass(c):
        print("class")
    @staticmethod
    def test():
        print("static method")
```

```python
p1 = Person1("a", 30)
p1.countClass()
```

```python
class Student(Person): #打上括号表示与Person建立关系
    def __init__(self,Name,age,stuId):
        super().__init__(Name,age) #super()表示父类，放哪里都可以
        print("This is prompted after super class")
        self.__ID = stuId
    def getstuID(self):
        return self.__ID
    def setstuID(self, id):
        self.__ID = id
    def Study(self, subject): 
        print(f"{Student.GetName()} studies {subject}")
    # overrided methods，不想用父类的方法时可以重新定义方法
    def move(self):
        super().move() # 如果要调用父类可以照常加super().
        print("Student move")
```

```python
class Teacher(Person): #打上括号表示与Person建立关系
    def __init__(self,Name,age,teacherId):
        super().__init__(Name,age) 
        self.__ID = teacherId
    def getTeacherId(self):
        return self.__ID
    def setTeacherId(self, id):
        self.__ID = id
    def Teach(self, subject):
        print(f"{Teacher.GetName()} teaches {subject}")
        
#子类之间不能互相调用
```

```python
p = Person("person", 29)
s = Student("student", 15,"09")
t = Teacher("Teacher", 34, "2018")

print(p.GetName())
print(s.GetName())

print()

s.move()

print()

p.move()
t.move()
```
