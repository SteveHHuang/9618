```python
class Cat:
    def __init__(self): # constructor method: create a new object/instance, it will be automatically called
        self.__name = "***"
        self.__weight = 10 # __ represent a private member
        self.__length = 10
        self.__blood = 10
    
    
    def eat(self, food):
        if self.__weight >= 40:
            print("cannot eat anymore")
        elif food == "fish":
            self.__weight +=5
            self.__blood += 10

        elif food == "meat":
            self.__weight +=2
            self.__blood += 4
        else:
            self.__weight+=1
            self.__blood +=2
        
    def assignName(self):
        self.__name = input("Enter the name")
        if self.__name == "":
            self.__name = "***"
            
        return self.__name
            
    def getWeight(self):
        if self.__weight >= 40:
            self.__weight = 40
        elif self.__weight<0:
            self.__weight = 0
            
        return self.__weight        
    
    def getBlood(self):
        if self.__blood >100:
            self.__blood += 10
        elif self.__blood<0:
            self.__blood = 0
            
        return self.__blood
        
```

```python
cat1 = Cat()
cat2 = Cat()
print(f"The blood of cat1 is {cat1.getBlood()}")
cat1.eat("fish")
print(f"The blood of cat1 is {cat1.getBlood()}")
print(f"The blood of cat2 is {cat2.getBlood()}")

cat2.eat("eggplant")
print(f"The blood of cat2 is {cat2.getBlood()}")
```
